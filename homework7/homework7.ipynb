```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 2. Matplotlib

def make_sin_wave(x, A, w):
    return A*np.sin(w*x)
x = np.linspace(0,2*np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

fig, ax = plt.subplots(1,3, figsize=(25,5))
# first subplot: varying amplitude
for A in amplitude:
    y = make_sin_wave(x, A, 1)
    ax[0].plot(x,y,label=f"A={A}")
ax[0].set_title("Varying Amplitude")
ax[0].set_xlabel("x")
ax[0].set_ylabel("y")
ax[0].legend(fontsize=12)

# second subplot: varying frequency
for w in frequency:
    y = make_sin_wave(x, 1, w)
    ax[1].plot(x,y, label = f"ω={w}")
ax[1].set_title("Varying Frequency")
ax[1].set_xlabel("x")
ax[1].set_ylabel("y")
ax[1].legend(fontsize=12)

# second subplot: varying both amplitude and frequency
for A,w in zip(amplitude, frequency):
    y = make_sin_wave(x, A, w,)
    ax[2].plot(x,y, label = f"A={A}, ω={w}")
ax[2].set_title("Varying Amplitude and Frequency")
ax[2].set_xlabel("x")
ax[2].set_ylabel("y")
ax[2].legend(fontsize=12)

plt.show()
```
