```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 3.1: Importing Data with NumPy
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)
print(data)
```

```python
# 3.2: Analyzing Imported Data
print(data.shape)
# there are 2642 rows.
magnitude = data[:,0]
temperature = data[:,1]

# minimums and maximums of each value with index
print(f"the minimum magnitude is {np.min(magnitude)} at index {np.argmin(magnitude)}")
print(f"the maximum magnitude is {np.max(magnitude)} at index {np.argmax(magnitude)}")
print(f"the minimum temperature is {np.min(temperature)} at index {np.argmin(temperature)}")
print(f"the maximum temperature is {np.max(temperature)} at index {np.argmax(temperature)}")
# star 1988 is brightest and star 264 is dimmest.
```

```python
# 3.3: Plot an HR Diagram
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
plt.figure(figsize = (8,6))
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
colors = ["blue"]*len(temperature)
plt.scatter(log_temperature[1988], log_magnitude[1988], c="black", s=150, label="Brightest Star")
plt.scatter(log_temperature[264], log_magnitude[264], c="purple", s=150, label="Dimmest Star")
plt.scatter(log_temperature[1923], log_magnitude[1923], c="green", s=150, label="Coolest Star")
plt.scatter(log_temperature[2613], log_magnitude[2613], c="orange", s=150, label="Hottest Star")
plt.scatter(log_temperature, log_magnitude, s=10, label="Other Stars", c=log_temperature, cmap="plasma")

plt.title("Hertzsprung-Russell Diagram")
plt.xlabel("Temperature (Log K)")
plt.ylabel("Magnitude")
plt.legend(fontsize = 12)
```
